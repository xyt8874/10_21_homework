```python
import torch

# 输入数据与对应的目标数据，均为浮点数
x = torch.tensor([-2.0, -1.5, -1.0, -0.5, 0.0, 0.5, 1.0, 1.5, 2.0])
y = torch.tensor([-3.0, -1.375, 0.0, 1.125, 2.0, 2.625, 3.0, 3.125, 3.0])
```

```python
w1 = torch.tensor(1.0, requires_grad=True)
w2 = torch.tensor(1.0, requires_grad=True)
w3 = torch.tensor(1.0, requires_grad=True)
y_pred = w1 * x + w2 * (x ** 2) + w3
loss = torch.abs(y - y_pred)
```

```python
w1 = torch.tensor(1.0, requires_grad=True)
w2 = torch.tensor(1.0, requires_grad=True)
w3 = torch.tensor(1.0, requires_grad=True)
optimizer = torch.optim.SGD([w1,w2,w3], lr=0.1)
epochs = 2000
```

```python
for epoch in range(epochs):
    y_pred = w1 * x + w2 * (x ** 2) + w3
    loss = torch.abs(y - y_pred).mean()
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print("=== 训练结束 ===")
print(f"学到的参数：w1 = {w1.item():.4f}, w2 = {w2.item():.4f}, w3 = {w3.item():.4f}")

with torch.no_grad():
    final_pred = w1 * x + w2 * (x ** 2) + w3
    final_loss = torch.abs(y - final_pred).mean()

print("\n每个样本的预测值：")
for i in range(len(x)):
    print(f"样本 {i+1}: 真实值 = {y[i].item():6.3f}, 预测值 = {final_pred[i].item():6.3f}")

print(f"\n最终的平均绝对误差 (Loss)：{final_loss.item():.6f}")
```

1.我的观察与说明:
我将参数w变为了三个参数，分别是w1,w2,w3。预测公式变为了y_pred = w1 * x + w2 * (x ** 2) + w3

2.我通过观察最终的平均绝对误差，只要loss相比初始状态显著下降，并且每个样本的预测值比较贴近真实值，就说明训练是有效的。
